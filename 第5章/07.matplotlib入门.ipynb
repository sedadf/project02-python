```python
import matplotlib.pyplot as plt

#展示中文
plt.rcParams['font.sans-serif'] = ['SimHei']

#绘制折线图
plt.figure(figsize=(8,5))#宽和高，单位是英寸
plt.plot([1,2,3,4,5],[1,4,9,16,25])

plt.title("气温折线图",fontsize=20)
plt.xlabel("天数",fontsize=14)
plt.ylabel("气温",fontsize=14)
plt.xticks([1,2,3,4,5])
plt.yticks([1,4,9,16,25])
plt.grid()
plt.legend(["气温"])


plt.show()

```
