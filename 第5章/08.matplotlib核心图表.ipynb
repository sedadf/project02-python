```python
from matplotlib.axes import Axes
import matplotlib.pyplot as plt
from IPython.core.pylabtools import figsize
from matplotlib.lines import lineStyles
from openai.resources.fine_tuning.alpha import alpha

#展示中文
plt.rcParams['font.sans-serif'] = ['SimHei']

#创建子图
#figure画布对象，axes子图数组
figure,axes=plt.subplots(nrows=1, ncols=2,figsize=(20,6),dpi=100)

#图一柱状图
countries=['中国','美国','英国','日本','俄罗斯','沙特阿拉伯','伊朗']#国家
values=[35,25,20,15,10,5,5]


axes1:Axes=axes[0]
axes1.bar(countries,values,color='r',width=0.5)
axes1.set_title("世界石油储备",fontsize=18)
axes1.set_xlabel("国家",fontsize=14)
axes1.set_ylabel("石油储备(亿吨)",fontsize=14)
axes1.grid(linestyle="--",alpha=0.3)

#图一饼状图
countries2 = ['印度', '中国', '美国', '印尼', '巴基斯坦', '尼日利亚', '巴西', '俄罗斯', '其他']
values = [14.51, 14.09, 3.4, 2.83, 2.51, 2.33, 2.12, 1.44, 20]
axes2:Axes=axes[1]
axes2.pie(values, labels=countries2, autopct='%1.1f%%')
axes2.legend(loc='lower center',ncol=5,bbox_to_anchor=(0.5,-0.1))

#保存图片
#plt.savefig('world_oil_reserves.png')




```
